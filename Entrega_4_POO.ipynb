```python
#Ejercicio pág 400
import tkinter as tk
from tkinter import scrolledtext


class Prueba:
    @staticmethod
    def correr():
        txt = ""

        try:
            txt += "Ingresando al primer try\n"
            div = 10000 / 0
            txt += "Después de la división\n"
        except ZeroDivisionError:
            txt += "División por cero\n"
        finally:
            txt += "Ingresando al primer finally\n"

        try:
            txt += "Ingresando al segundo try\n"
            obj = None
            obj.to_string()
            txt += "Imprimiendo objeto\n"
        except ZeroDivisionError:
            txt += "División por cero\n"
        except Exception:
            txt += "Ocurrió una excepción\n"
        finally:
            txt += "Ingresando al segundo finally\n"

        return txt


class FueraLimite:
    @staticmethod
    def correr():
        txt = ""

        try:
            cad = "Programación"
            car = cad[14]
            txt += car + "\n"
        except IndexError:
            txt += "Índice de string por fuera del límite\n"

        return txt


class FormatoNum:
    @staticmethod
    def correr():
        txt = ""

        try:
            num = int("Número")
            txt += str(num) + "\n"
        except ValueError:
            txt += "Excepción de formato de número\n"
        finally:
            txt += "Ingresando al finally\n"

        return txt


class Ventana:
    def __init__(self):
        self.v = tk.Tk()
        self.v.title("Manejo de excepciones")
        self.v.geometry("620x430")
        self.v.resizable(False, False)

        self.armar()

    def armar(self):
        lbl_titulo = tk.Label(
            self.v,
            text="Manejo de excepciones",
            font=("Arial", 16, "bold")
        )
        lbl_titulo.pack(pady=10)

        zona_btn = tk.Frame(self.v)
        zona_btn.pack(pady=10)

        btn1 = tk.Button(
            zona_btn,
            text="Prueba principal",
            width=28,
            command=self.probar
        )
        btn1.grid(row=0, column=0, padx=5, pady=5)

        btn2 = tk.Button(
            zona_btn,
            text="Fuera de límite",
            width=28,
            command=self.probar_limite
        )
        btn2.grid(row=0, column=1, padx=5, pady=5)

        btn3 = tk.Button(
            zona_btn,
            text="Formato de número",
            width=28,
            command=self.probar_numero
        )
        btn3.grid(row=1, column=0, padx=5, pady=5)

        btn4 = tk.Button(
            zona_btn,
            text="Limpiar",
            width=28,
            command=self.limpiar
        )
        btn4.grid(row=1, column=1, padx=5, pady=5)

        lbl_res = tk.Label(
            self.v,
            text="Salida del programa:",
            font=("Arial", 11, "bold")
        )
        lbl_res.pack(pady=5)

        self.salida = scrolledtext.ScrolledText(
            self.v,
            width=70,
            height=13,
            font=("Consolas", 10)
        )
        self.salida.pack(pady=5)

    def poner(self, titulo, texto):
        self.salida.insert(tk.END, f"--- {titulo} ---\n")
        self.salida.insert(tk.END, texto)
        self.salida.insert(tk.END, "\n")

    def probar(self):
        r = Prueba.correr()
        self.poner("Prueba principal", r)

    def probar_limite(self):
        r = FueraLimite.correr()
        self.poner("Fuera de límite", r)

    def probar_numero(self):
        r = FormatoNum.correr()
        self.poner("Formato de número", r)

    def limpiar(self):
        self.salida.delete("1.0", tk.END)

    def iniciar(self):
        self.v.mainloop()


if __name__ == "__main__":
    app = Ventana()
    app.iniciar()
```

```python
#Ejercicio pág 406
import tkinter as tk
from tkinter import messagebox, scrolledtext


class Vendedor:
    def __init__(self, nom, ape):
        self.nom = nom
        self.ape = ape
        self.edad = 0

    def imprimir(self):
        txt = ""
        txt += f"Nombre del vendedor = {self.nom}\n"
        txt += f"Apellidos del vendedor = {self.ape}\n"
        txt += f"Edad del vendedor = {self.edad}\n"
        return txt

    def revisar_edad(self, ed):
        if ed < 18:
            raise ValueError("El vendedor debe ser mayor de 18 años.")
        elif ed > 0 and ed < 120:
            self.edad = ed
        else:
            raise ValueError("La edad no puede ser negativa ni mayor a 120.")


class Ventana:
    def __init__(self):
        self.v = tk.Tk()
        self.v.title("Registro de vendedor")
        self.v.geometry("600x430")
        self.v.resizable(False, False)

        self.armar()

    def armar(self):
        titulo = tk.Label(
            self.v,
            text="Registro de vendedor",
            font=("Arial", 16, "bold")
        )
        titulo.pack(pady=10)

        marco = tk.Frame(self.v)
        marco.pack(pady=10)

        tk.Label(marco, text="Nombre:").grid(row=0, column=0, padx=8, pady=8, sticky="e")
        self.c_nom = tk.Entry(marco, width=30)
        self.c_nom.grid(row=0, column=1, padx=8, pady=8)

        tk.Label(marco, text="Apellidos:").grid(row=1, column=0, padx=8, pady=8, sticky="e")
        self.c_ape = tk.Entry(marco, width=30)
        self.c_ape.grid(row=1, column=1, padx=8, pady=8)

        tk.Label(marco, text="Edad:").grid(row=2, column=0, padx=8, pady=8, sticky="e")
        self.c_edad = tk.Entry(marco, width=30)
        self.c_edad.grid(row=2, column=1, padx=8, pady=8)

        zona_btn = tk.Frame(self.v)
        zona_btn.pack(pady=10)

        btn1 = tk.Button(
            zona_btn,
            text="Registrar",
            width=20,
            command=self.registrar
        )
        btn1.grid(row=0, column=0, padx=8)

        btn2 = tk.Button(
            zona_btn,
            text="Limpiar",
            width=20,
            command=self.limpiar
        )
        btn2.grid(row=0, column=1, padx=8)

        tk.Label(
            self.v,
            text="Salida del programa:",
            font=("Arial", 11, "bold")
        ).pack(pady=5)

        self.salida = scrolledtext.ScrolledText(
            self.v,
            width=65,
            height=12,
            font=("Consolas", 10)
        )
        self.salida.pack(pady=5)

    def registrar(self):
        self.salida.delete("1.0", tk.END)

        nom = self.c_nom.get().strip()
        ape = self.c_ape.get().strip()
        ed_txt = self.c_edad.get().strip()

        try:
            if nom == "" or ape == "" or ed_txt == "":
                raise ValueError("Todos los campos son obligatorios.")

            ed = int(ed_txt)

            ven = Vendedor(nom, ape)
            ven.revisar_edad(ed)

            self.salida.insert(tk.END, ven.imprimir())

        except ValueError as e:
            self.salida.insert(tk.END, f"{e}\n")
            messagebox.showerror("Error", str(e))

    def limpiar(self):
        self.c_nom.delete(0, tk.END)
        self.c_ape.delete(0, tk.END)
        self.c_edad.delete(0, tk.END)
        self.salida.delete("1.0", tk.END)

    def iniciar(self):
        self.v.mainloop()


if __name__ == "__main__":
    app = Ventana()
    app.iniciar()
```

```python
#ejercicio pág 412
import math
import tkinter as tk

class Calculos:
    @staticmethod
    def calcular_logaritmo_neperiano(numero):
        if numero <= 0:
            raise ValueError("El numero debe ser positivo")
        else:
            logaritmo = math.log(numero)

        return logaritmo

    @staticmethod
    def calcular_raiz_cuadrada(numero):
        if numero <= 0:
            raise ValueError("El numero debe ser positivo")
        else:
            raiz = math.sqrt(numero)

        return raiz

class Ventana:
    def __init__(self):

        self.ventana = tk.Tk()
        self.ventana.title("Ejercicio_pag412")
        self.ventana.geometry("500x400")
        self.ventana.columnconfigure(0, weight=1)

        self.titulo = tk.Label(self.ventana, text="Calculos Númericos", font=("Arial", 12, "bold"))
        self.titulo.grid(row=0, column=0, columnspan=2, pady=15)

        self.texto = tk.Label(self.ventana, text="Ingrese el numero:", font=("Arial", 11))
        self.texto.grid(row=1, column=0, pady=10)

        self.entrada = tk.Entry(self.ventana, width=20, font=("Arial", 11))
        self.entrada.grid(row=2, column=0, pady=10)

        self.boton = tk.Button(self.ventana, text="Calcular", command=self.hacer_calculos)
        self.boton.grid(row=3, column=0, pady=15)


        self.mostrar_logartimo = tk.Label(self.ventana, text="")
        self.mostrar_logartimo.grid(row=4, column=0, pady=10)


        self.mostrar_raiz = tk.Label(self.ventana, text="")
        self.mostrar_raiz.grid(row=5, column=0, pady=10)


        self.ventana.mainloop()


    def hacer_calculos(self):
        try:
            self.numero = float(self.entrada.get())
            self.entrada.delete(0, tk.END)


            self.mostrar_logartimo.config(text=f"El logartimo es: {Calculos.calcular_logaritmo_neperiano(self.numero):.2f}")
            self.mostrar_raiz.config(text=f"La raiz es: {Calculos.calcular_raiz_cuadrada(self.numero):.2f}")

        except ValueError as error_mensaje:
            self.mostrar_logartimo.config(text=f"¡Error! {error_mensaje}")
            self.mostrar_raiz.config(text="")

app = Ventana()
```

```python
#Ejercicio pág 418
 import tkinter as tk

class EquipoMaratonProgramacion:
    def __init__(self, nombre_equipo, university, lenguaje):
        self.nombre_equipo = nombre_equipo
        self.universidad = university
        self.lenguaje_programacion = lenguaje
        self.programadores = []

    def esta_completo(self):
        if len(self.programadores) >= 2:
            return True
        else:
            return False

    @staticmethod
    def validar_nombre(nombre_completo):
        if nombre_completo == "":
            raise ValueError("El nombre no puede estar vacío")

        if len(nombre_completo) >= 20:
            raise ValueError("El nombre no puede tener más de 20 caracteres")

        if any(char.isdigit() for char in nombre_completo):
            raise ValueError("El nombre no puede contener números")

        return True

    def añadir_programador(self, nombre_completo):
        if len(self.programadores) >= 3:
            raise ValueError("Este grupo ya está lleno")

        self.programadores.append(nombre_completo)

    def elementos_equipo(self):
        nombres_integrantes = "\n".join([f"  • {i}" for i in self.programadores])

        elementos = f"Elementos del equipo\n" \
                    f"Equipo: {self.nombre_equipo}\n" \
                    f"Universidad: {self.universidad}\n" \
                    f"Lenguaje: {self.lenguaje_programacion}\n" \
                    f"Integrantes ({len(self.programadores)}):\n{nombres_integrantes}"
        return elementos

class Ventana:
    def __init__(self):
        self.ventana = tk.Tk()
        self.ventana.title("Ejercicio_pag418")
        self.ventana.geometry("500x550")

        self.titulo = tk.Label(self.ventana, text="Maratón Programación", font=("Arial", 12, "bold"))
        self.titulo.pack(pady=10)

        self.equipo = None

        tk.Label(self.ventana, text="REGISTRO DEL EQUIPO", font=("Arial", 11, "bold")).pack(pady=5)

        self.bloque_equipo = tk.Frame(self.ventana)
        self.bloque_equipo.pack(pady=5)

        tk.Label(self.bloque_equipo, text="Nombre Equipo:").grid(row=0, column=0, sticky="e", pady=5)
        self.entrada_nombre_equipo = tk.Entry(self.bloque_equipo, width=25)
        self.entrada_nombre_equipo.grid(row=0, column=1, padx=10, pady=5)

        tk.Label(self.bloque_equipo, text="Universidad:").grid(row=1, column=0, sticky="e", pady=5)
        self.entrada_universidad = tk.Entry(self.bloque_equipo, width=25)
        self.entrada_universidad.grid(row=1, column=1, padx=10, pady=5)

        tk.Label(self.bloque_equipo, text="Lenguaje:").grid(row=2, column=0, sticky="e", pady=5)
        self.entrada_lenguaje = tk.Entry(self.bloque_equipo, width=25)
        self.entrada_lenguaje.grid(row=2, column=1, padx=10, pady=5)

        self.btn_crear_equipo = tk.Button(self.ventana, text="Crear Equipo", command=self.crear_equipo_interfaz)
        self.btn_crear_equipo.pack(pady=10)

        self.label_programadores = tk.Label(self.ventana, text="INTEGRANTES DEL EQUIPO", font=("Arial", 11, "bold"), fg="gray")
        self.label_programadores.pack(pady=10)

        self.bloque_programadores = tk.Frame(self.ventana)
        self.bloque_programadores.pack(pady=5)

        tk.Label(self.bloque_programadores, text="Nombre y Apellido:").grid(row=0, column=0, sticky="e", pady=5)
        self.entrada_programadores = tk.Entry(self.bloque_programadores, width=25, state="disabled")
        self.entrada_programadores.grid(row=0, column=1, padx=10, pady=5)

        self.btn_añadir_prog = tk.Button(self.ventana, text="Añadir Programador", command=self.añadir_programador_interfaz, state="disabled")
        self.btn_añadir_prog.pack(pady=10)

        self.label_lista = tk.Label(self.ventana, text="Lista de integrantes:\n(Vacía)", font=("Arial", 10, "italic"))
        self.label_lista.pack(pady=15)

        self.btn_elementos = tk.Button(self.ventana, text="Mostrar Elementos", command=self.mostrar_elementos_interfaz, state="disabled")
        self.btn_elementos.pack(pady=10)

        self.ventana.mainloop()


    def crear_equipo_interfaz(self):
        nombre_equipo = self.entrada_nombre_equipo.get().strip()
        universidad = self.entrada_universidad.get().strip()
        lenguaje = self.entrada_lenguaje.get().strip()

        if nombre_equipo == "" or universidad == "" or lenguaje == "":
            self.label_lista.config(text="¡Error! Por favor llena todos los campos", font=("Arial", 10, "bold"), fg="red")
            return

        self.equipo = EquipoMaratonProgramacion(nombre_equipo, universidad, lenguaje)

        self.entrada_nombre_equipo.config(state="disabled")
        self.entrada_universidad.config(state="disabled")
        self.entrada_lenguaje.config(state="disabled")
        self.btn_crear_equipo.config(state="disabled")

        self.entrada_programadores.config(state="normal")
        self.btn_añadir_prog.config(state="normal")
        self.label_programadores.config(fg="black")

        self.label_lista.config(text="Equipo creado con éxito. Agrega sus integrantes.", font=("Arial", 10, "italic"), fg="black")


    def añadir_programador_interfaz(self):
        nombre_p = self.entrada_programadores.get().strip()

        try:

            EquipoMaratonProgramacion.validar_nombre(nombre_p)
            self.equipo.añadir_programador(nombre_p)

            self.entrada_programadores.delete(0, tk.END)

            texto_lista = f"Lista de integrantes ({len(self.equipo.programadores)}/3):\n" + "\n".join([f"- {i}" for i in self.equipo.programadores])
            self.label_lista.config(text=texto_lista, font=("Arial", 10, "normal"), fg="black")

            if self.equipo.esta_completo():
                self.btn_elementos.config(state="normal")

        except ValueError as error:
            self.label_lista.config(text=f"¡Error! {error}", font=("Arial", 10, "bold"), fg="red")


    def mostrar_elementos_interfaz(self):
        info_completa = self.equipo.elementos_equipo()
        self.label_lista.config(text=info_completa, font=("Arial", 10, "bold"), justify="left", fg="black")


app = Ventana()
```

```python
#Ejercicio 427
import tkinter as tk

class LeerArchivo:
    @staticmethod
    def leer_y_mostrar(nombre_archivo, componente_texto):
        componente_texto.delete("1.0", tk.END)

        archivo = None
        try:
            archivo = open(nombre_archivo, mode='r', encoding='utf-8')

            linea = archivo.readline()

            while linea != "":
                componente_texto.insert(tk.END, linea)

                linea = archivo.readline()

            archivo.close()

        except FileNotFoundError:
            componente_texto.insert(tk.END, f"El archivo '{nombre_archivo}' no fue encontrado.")
        except Exception as e:
            componente_texto.insert(tk.END, f"¡Error de Lectura! {e}")


class Ventana:
    def __init__(self):
        self.ventana = tk.Tk()
        self.ventana.title("Ejercicio_Pag427")
        self.ventana.geometry("500x450")

        self.titulo = tk.Label(self.ventana, text="Lector de Archivos (.txt)", font=("Arial", 12, "bold"))
        self.titulo.pack(pady=10)

        self.bloque_ruta = tk.Frame(self.ventana)
        self.bloque_ruta.pack(pady=5)

        tk.Label(self.bloque_ruta, text="Ruta del archivo:").grid(row=0, column=0, padx=5)

        self.entrada_ruta = tk.Entry(self.bloque_ruta, width=35)
        self.entrada_ruta.insert(0, "C:/prueba.txt")
        self.entrada_ruta.grid(row=0, column=1, padx=5)

        self.btn_leer = tk.Button(self.ventana, text="Leer Contenido", command=self.ejecutar_lectura)
        self.btn_leer.pack(pady=10)

        tk.Label(self.ventana, text="Contenido del archivo:", font=("Arial", 10, "bold")).pack(pady=5)

        self.caja_contenido = tk.Text(self.ventana, width=55, height=12, font=("Courier", 10))
        self.caja_contenido.pack(pady=5)

        self.ventana.mainloop()

    def ejecutar_lectura(self):
        ruta = self.entrada_ruta.get().strip()

        LeerArchivo.leer_y_mostrar(ruta, self.caja_contenido)

app = Ventana()
```
